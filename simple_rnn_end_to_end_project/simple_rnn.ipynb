```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
## Load the IMDB dataset
max_features = 10000  # Number of unique words to consider
(X_train, y_train),(X_test, y_test) = imdb.load_data(num_words = max_features  )

print("Training data shape:", X_train.shape)
print("Test data shape:", X_test.shape)
```

```python
sample_review = X_train[0]
sample_label = y_train[0]
print("Sample review (encoded):", sample_review)
print("Sample label (0=negative, 1=positive):", sample_label)
```

```python
word_index = imdb.get_word_index()
rev_word_index= {value : key for key , value in word_index.items()}
```

```python
decoded_review = ' '.join([rev_word_index.get(i - 3, '?') for i in sample_review])
decoded_review
```

```python
from tensorflow.keras.preprocessing import sequence
maxlen = 500  # Maximum length of each review
X_train = sequence.pad_sequences(X_train, maxlen=maxlen)  
X_test = sequence.pad_sequences(X_test, maxlen=maxlen)
print("Padded training data shape:", X_train.shape)   
```

```python
model = Sequential()
model.add(Embedding(max_features,128, input_length=maxlen))  # Embedding layer
model.add(SimpleRNN(128,activation='relu'))  # Simple RNN layer
model.add(Dense(1, activation='sigmoid'))  # Output layer for binary classification  

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
early_stooping = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
```

```python
model_history = model.fit(
    X_train,y_train, epochs =10, batch_size=32, 
    validation_split=0.2, callbacks=[early_stooping]
)
```

```python
#save the model
model.save('simple_rnn_model_imdb.h5')
```
