```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
## Define vocabulary size
voc_size = 10000
### One hot representation 
one_hot_repr = [one_hot(word, voc_size) for word in sent]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential

```

```python
import numpy as np
```

```python
sent_lenth = 8 # max length of sentence
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_lenth)
print(embedded_docs)
```

```python
# feature representation
dim = 10 
model = Sequential()
model.add(Embedding(voc_size,dim,input_length = sent_lenth))
model.compile('adam', 'mse')
print(model.summary())

```

```python
model.predict(embedded_docs )
```
